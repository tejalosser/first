```python
import pandas as pd
df=pd.read_csv('/content/amazon.csv')
df
```

```python
df.drop_duplicates()
```

```python
df.iloc[0,1]
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
vectorizer=TfidfVectorizer(stop_words='english')
X=vectorizer.fit_transform(df['product_name'])
X.toarray()
```

```python
def similar(product):
  y=vectorizer.transform(product)
  similarity=cosine_similarity(X,y)
  similarity=similarity.flatten()
  idx=[]
  for sim in similarity:
    if sim > 0.5 and sim != 1.0:
      idx.append(similarity.tolist().index(sim))
  return idx
```

```python
df_product=df['product_name'].head(10)
df_product.head(10)
```

```python
df_product[3]
```

```python

List=similar(pd.Series(df_product[3]))
print(List)
for i in List:
  print(df.loc[i,'product_name'])
```

```python
import numpy as np
import pandas as pd
df1=pd.read_csv('/content/ratings_Electronics (1).csv')
df1.head()
```

```python
len(df1)
```

```python
df1=df1.iloc[:10000,]
df1.head()
```

```python
df1.columns=['userid','productid','rating','timestamp']
```

```python
df1.drop('timestamp',axis=1,inplace=True)
```

```python
df1.head()
```

```python
product_counts =df1.groupby('productid',as_index=False)['rating'].agg(['size','mean'])
product_counts
```

```python
product_counts=product_counts[(product_counts['size']>=100) & (product_counts['mean']>3)]
product_counts.sort_values(by='mean',ascending=False)
```
